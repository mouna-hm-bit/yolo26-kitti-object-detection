```python

from ultralytics import YOLO

# ==========================================
# 1. LOAD COCO PRETRAINED YOLO26
# ==========================================

model = YOLO("yolo26n.pt")

# COCO pretrained model: 80 classes
print("Pretrained model loaded!")

# ==========================================
# 2. TRANSFER LEARNING ON KITTI
# ==========================================

results = model.train(
    data="kitti.yaml",       # KITTI dataset configuration
    epochs=10,
    imgsz=640,
    batch=128,
    device="0,1",            # Two T4 GPUs
    amp=True,                # Automatic Mixed Precision
    patience=20,
    workers=8,
    project="KITTI_Project",
    name="YOLO26n_Transfer_Learning"
)


```

```python
# ==========================================
# 3. LOAD THE BEST TRAINED MODEL
# ==========================================

best_model = YOLO(
    "/kaggle/working/runs/detect/KITTI_Project/YOLO26n_Transfer_Learning/weights/best.pt"
)

# Check the number of classes
print("Number of classes:", best_model.model.nc)
print("KITTI classes:", best_model.names)
```
